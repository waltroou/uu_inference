## Problem 1

A random sample has
   $$
   \bar{X}=52,\qquad s=10,\qquad n=100.
   $$
   Compute the estimated standard error of $\bar{X}$. Test
   $$
   H_0:\mu=50
   \qquad \text{vs.} \qquad
   H_1:\mu\ne 50
   $$
   at significance levels $\alpha=.10,.05,.01$ using the large-sample $z$-statistic. These correspond to 90%, 95%, and 99% confidence levels. Construct a 95% confidence interval for $\mu$.

```python
sample_mean = 52
sample_sd = 10
n = 100
null_mean = 50

se = sample_sd / np.sqrt(n)
z = (sample_mean - null_mean) / se

print('Standard error:', se)
print('z-statistic:', z)
print('Reject at 10%:', abs(z) > z90)
print('Reject at 5%:', abs(z) > z95)
print('Reject at 1%:', abs(z) > z99)

lower = sample_mean - z95 * se
upper = sample_mean + z95 * se
print('95% confidence interval:', lower, 'to', upper)
```

## Problem 2

Suppose
   $$
   \hat{S}=3.1,\qquad \widehat{SE}(\hat{S})=1.3.
   $$
   Construct 90%, 95%, and 99% confidence intervals, and test the null hypothesis that $S=0$ at the corresponding significance levels $\alpha=.10,.05,.01$. When do you reject or fail to reject the null of $S=0$?

```python
estimate = 3.1
se = 1.3

print('90% CI:', estimate - 1.645 * se, 'to', estimate + 1.645 * se)
print('95% CI:', estimate - 1.96 * se, 'to', estimate + 1.96 * se)
print('99% CI:', estimate - 2.576 * se, 'to', estimate + 2.576 * se)

z = estimate / se
print('z-statistic:', z)
```


- answer: Reject S = 0 at 10% and 5% because those intervals exclude zero. Fail to reject at 1% because the 99% interval includes zero. 

## Problem 3

A health survey compares a biomarker to a benchmark value of 12. The sample has
   $$
   \bar{X}=11.5,\qquad s=3,\qquad n=36.
   $$
   Is 12 a plausible value for the population mean? Construct a 95% confidence interval and compute the two-sided p-value for $H_0:\mu=12$ versus $H_1:\mu\ne12$. Assess plausibility using the interval and the test at $\alpha=.05$.

```python
sample_mean = 11.5
sample_sd = 3
n = 36
null_mean = 12

se = sample_sd / np.sqrt(n)
z = (sample_mean - null_mean) / se
p_value = 2 * norm.cdf(-abs(z))

lower = sample_mean - 1.96 * se
upper = sample_mean + 1.96 * se

print('Standard error:', se)
print('z-statistic:', z)
print('95% CI:', lower, 'to', upper)
print('Two-sided p-value:', p_value)
```


- answer: The two-sided p-value is about 0.317. Since it is greater than 0.05, we fail to reject the null. 12 is plausible because it is inside the interval, and the test does not show a clear difference from 12.

## Problem 4

A real estate analyst compares the average price of a house in two neighborhoods.
    Group A:
    $$
    \bar{X}_A=25,\qquad s_A=8,\qquad n_A=100
    $$
    Group B:
    $$
    \bar{X}_B=22,\qquad s_B=7,\qquad n_B=100.
    $$
    Compute the two-sample $Z$-statistic for testing equality of means. Construct an approximate 95% confidence interval for $\mu_A-\mu_B$.

```python
difference = 25 - 22
se = np.sqrt(8**2 / 100 + 7**2 / 100)
z = difference / se
print('z:', z)
print('95% CI:', difference - 1.96 * se, 'to', difference + 1.96 * se)
```

- answer: The z-statistic is about 2.82. The 95% interval for A minus B is [0.92, 5.08]. It excludes zero, giving evidence of different mean prices at the 5% level.

## Problem 7

For `./data/ames_prices.csv`, define an indicator equal to 1 when `Central.Air == 'Y'`. Estimate the proportion of homes with central air and test $H_0:p=.90$. Treat a proportion as the mean of a binary variable and complete the full inference routine.

```python
ames = pd.read_csv('./data/ames_prices.csv')
air = ames['Central.Air'].dropna()
data = (air == 'Y').astype(int).to_numpy()
estimate = data.mean()
null = 0.90
seed = 47
rng = np.random.default_rng(seed)
R = 10000
boot = []
for i in range(R):
    sample = rng.choice(data, size=len(data), replace=True)
    boot.append(sample.mean())

se = np.std(boot, ddof=1)
z = (estimate - null) / se
p_value = 2 * norm.cdf(-abs(z))
print('Seed:', seed, 'Bootstrap samples:', R)
print('Sample size:', len(data))
print('Estimate:', estimate, 'Standard error:', se)
print('z:', z, 'p-value:', p_value)

for confidence, critical, alpha in [(90, 1.645, 0.10), (95, 1.96, 0.05), (99, 2.576, 0.01)]:
    lower = estimate - critical * se
    upper = estimate + critical * se
    decision = 'Reject' if p_value < alpha else 'Fail to reject'
    agrees = (p_value < alpha) == (null < lower or null > upper)
    print(confidence, '% CI:', round(lower, 4), 'to', round(upper, 4),
          '|', decision, '| Test and CI agree:', agrees)
```

- answer: About 93.3% of homes have central air. The 95% interval is 92.4% to 94.2%. Reject 90% at all three levels; the data support a higher proportion. The output gives all intervals and confirms the decisions agree.

```python
data = ames[['area', 'price']].dropna()
data = np.arcsinh(data).to_numpy()
estimate = np.corrcoef(data[:, 0], data[:, 1])[0, 1]
null = 0
seed = 48
rng = np.random.default_rng(seed)
R = 10000
boot = []
for i in range(R):
    rows = rng.choice(len(data), size=len(data), replace=True)
    sample = data[rows]
    boot.append(np.corrcoef(sample[:, 0], sample[:, 1])[0, 1])

se = np.std(boot, ddof=1)
z = (estimate - null) / se
p_value = 2 * norm.cdf(-abs(z))
print('Seed:', seed, 'Bootstrap samples:', R)
print('Sample size:', len(data))
print('Estimate:', estimate, 'Standard error:', se)
print('z:', z, 'p-value:', p_value)

for confidence, critical, alpha in [(90, 1.645, 0.10), (95, 1.96, 0.05), (99, 2.576, 0.01)]:
    lower = estimate - critical * se
    upper = estimate + critical * se
    decision = 'Reject' if p_value < alpha else 'Fail to reject'
    agrees = (p_value < alpha) == (null < lower or null > upper)
    print(confidence, '% CI:', round(lower, 4), 'to', round(upper, 4),
          '|', decision, '| Test and CI agree:', agrees)
```

- answer: The correlation is about 0.723, with a 95% interval of [0.704, 0.743]. Reject zero correlation at all three levels. Larger area is associated with higher price, but this does not prove cause and effect. Resampling area and price separately would break the pairs and lose their relationship. All decisions agree with the intervals. The printed p-value is 0 because it is too small for the computer to represent, not because it is exactly zero.
