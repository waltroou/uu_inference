## Problem 1

```python
xbar, s, n, mu0 = 52, 10, 100, 50

se = s / np.sqrt(n)
z = (xbar - mu0) / se

print("SE:", se)
print("z:", z)

for level, z_star in [("90%", 1.645), ("95%", 1.96), ("99%", 2.576)]:
    ci_low = xbar - z_star * se
    ci_high = xbar + z_star * se
    print(level, "CI:", ci_low, ci_high, "| contains 50:", ci_low <= mu0 <= ci_high)
```

- Reject at 10% and 5% (both intervals exclude 50), fail to reject at 1% (the 99% interval contains it)

## Problem 3

```python
xbar, s, n, mu0 = 11.5, 3, 36, 12

se = s / np.sqrt(n)
ci_low = xbar - 1.96 * se
ci_high = xbar + 1.96 * se

print("SE:", se)
print("95% CI:", round(ci_low, 4), round(ci_high, 4), "| contains 12:", ci_low <= mu0 <= ci_high)
```

- H0: mu = 12 vs H1: mu != 12 at alpha = .05: fail to reject, since the 95% interval contains 12.
- 12 is therefore a plausible value for the population mean. The sample mean of 11.5 sits one standard error below the benchmark, which is ordinary sampling variation.

## Problem 7

```python
ames = pd.read_csv("./data/ames_prices.csv")
central_air = (ames["Central.Air"] == "Y").astype(int).values

seed = 42
R = 2000
rng = np.random.default_rng(seed)
n = len(central_air)
theta_hat = central_air.mean()
boot = np.array([central_air[rng.integers(0, n, n)].mean() for _ in range(R)])
theta0 = 0.90

se_hat = boot.std(ddof=1)
print("seed:", seed, "R:", R)
print("estimate:", theta_hat)
print("SE:", se_hat)
for level, z_star in [("90%", 1.645), ("95%", 1.96), ("99%", 2.576)]:
    ci_low = theta_hat - z_star * se_hat
    ci_high = theta_hat + z_star * se_hat
    print(level, "CI:", round(ci_low, 4), round(ci_high, 4), "| contains 0.90:", ci_low <= theta0 <= ci_high)
```

- About 93% of these homes have central air, comfortably above the 90% benchmark. All three intervals exclude .90, so H0: p = .90 is rejected at 10%, 5% and 1%, and the evidence is strong rather than borderline.


## Problem 8

```python
ames = pd.read_csv("./data/ames_prices.csv")
x = np.arcsinh(ames["area"]).values
y = np.arcsinh(ames["price"]).values

seed = 42
R = 2000
rng = np.random.default_rng(seed)
n = len(x)
theta_hat = np.corrcoef(x, y)[0, 1]
rows = np.array([rng.integers(0, n, n) for _ in range(R)])
boot = np.array([np.corrcoef(x[i], y[i])[0, 1] for i in rows])
theta0 = 0.0

se_hat = boot.std(ddof=1)
print("seed:", seed, "R:", R)
print("estimate:", theta_hat)
print("SE:", se_hat)
for level, z_star in [("90%", 1.645), ("95%", 1.96), ("99%", 2.576)]:
    ci_low = theta_hat - z_star * se_hat
    ci_high = theta_hat + z_star * se_hat
    print(level, "CI:", round(ci_low, 4), round(ci_high, 4), "| contains 0:", ci_low <= theta0 <= ci_high)
```

- The correlation is strong and positive: all three intervals sit around 0.70 to 0.75 and exclude 0, so H0: rho = 0 is rejected at 10%, 5% and 1%. Homes with more interior area tend to sell for more; the association is not evidence that size causes price.
- Resampling area and price separately would break the pairing: each replicate would match a resampled area with an unrelated resampled price, so the bootstrap distribution would collapse toward 0 and its spread would describe two independent variables instead of the sampling variability of rho.

## Problem 9

```python
metabric = pd.read_csv("./data/metabric.csv")
months = metabric["Overall Survival (Months)"].values

seed = 42
R = 2000
rng = np.random.default_rng(seed)
n = len(months)
theta_hat = months.mean()
boot = np.array([months[rng.integers(0, n, n)].mean() for _ in range(R)])
theta0 = 120

se_hat = boot.std(ddof=1)
print("seed:", seed, "R:", R)
print("estimate:", theta_hat)
print("SE:", se_hat)
for level, z_star in [("90%", 1.645), ("95%", 1.96), ("99%", 2.576)]:
    ci_low = theta_hat - z_star * se_hat
    ci_high = theta_hat + z_star * se_hat
    print(level, "CI:", round(ci_low, 2), round(ci_high, 2), "months |",
          round(ci_low / 12, 2), round(ci_high / 12, 2), "years | contains 120:", ci_low <= theta0 <= ci_high)
```

- The average recorded time is about 129 months, roughly 9 months above the 120-month benchmark, and all three intervals exclude 120, so H0: mu = 120 is rejected at 10%, 5% and 1%. In years the same intervals run from about 10.3 to 11.2.
- This does not estimate unrestricted mean lifetime. Patients still alive at last contact are right-censored: their recorded time stops at follow-up, which is shorter than their eventual survival time, so the mean of recorded times understates the cohort's mean survival.

## Problem 10

```python
nhanes = pd.read_csv("./data/nhanes_data_17_18.csv", low_memory=False)
adults = nhanes[(nhanes["AgeInYearsAtScreening"] >= 18) & nhanes["SleepHoursWeekdaysOrWorkdays"].notna()]
sleep = adults["SleepHoursWeekdaysOrWorkdays"].values

seed = 42
R = 2000
rng = np.random.default_rng(seed)
n = len(sleep)
theta_hat = sleep.mean()
boot = np.array([sleep[rng.integers(0, n, n)].mean() for _ in range(R)])
theta0 = 8.0

se_hat = boot.std(ddof=1)
print("seed:", seed, "R:", R)
print("estimate:", theta_hat)
print("n:", n)
print("SE:", se_hat)
for level, z_star in [("90%", 1.645), ("95%", 1.96), ("99%", 2.576)]:
    ci_low = theta_hat - z_star * se_hat
    ci_high = theta_hat + z_star * se_hat
    print(level, "CI:", round(ci_low, 4), round(ci_high, 4), "| contains 8.0:", ci_low <= theta0 <= ci_high)
```

- Mean weekday sleep is about 7.62 hours, roughly 23 minutes below the 8-hour benchmark, and all three intervals exclude 8, so H0: mu = 8 is rejected at 10%, 5% and 1%.
- Statistical and practical significance tell different stories here: with n = 5489 even a gap this small is resolved, but 23 minutes a night is a modest difference in practical terms.
